Assignment 1: Calculate State Value and Action Value for each state

grid_env_det.py

```python
%%writefile grid_env_det.py

state_space = ["S0", "S1", "S3", "S4", "G"]

action_space = ["up", "down", "left", "right"]

discount_factor = 0.9

reward_function = {
    "normal": -1,
    "goal": 10,
    "obstacle": -5
}

transition_function = {
    ("S0", "up"): "S0",
    ("S0", "down"): "S0",
    ("S0", "left"): "S0",
    ("S0", "right"): "S1",

    ("S1", "up"): "S1",
    ("S1", "down"): "S3",
    ("S1", "left"): "S0",
    ("S1", "right"): "G",

    ("S3", "up"): "S1",
    ("S3", "down"): "S3",
    ("S3", "left"): "X",
    ("S3", "right"): "S4",

    ("S4", "up"): "G",
    ("S4", "down"): "S4",
    ("S4", "left"): "S3",
    ("S4", "right"): "S4"
}

def transition(state, action):
    if state == "G":
        return "G", reward_function["goal"], True

    next_state = transition_function[(state, action)]

    if next_state == "X":
        return "X", reward_function["obstacle"], True

    if next_state == "G":
        return "G", reward_function["goal"], True

    return next_state, reward_function["normal"], False
```

```python
import grid_env_det

policy = {
    "S0": "right",
    "S1": "down",
    "S3": "right",
    "S4": "up"
}

def state_value_function(state):
    total_return = 0
    gamma = grid_env_det.discount_factor
    current_state = state
    power = 0

    while current_state != "G":
        action = policy[current_state]
        next_state, reward, done = grid_env_det.transition(current_state, action)

        total_return += (gamma ** power) * reward
        power += 1
        current_state = next_state

    return total_return

def action_value_function(state, action):
    total_return = 0
    gamma = grid_env_det.discount_factor
    current_state = state
    power = 0

    while current_state != "G":
        next_state, reward, done = grid_env_det.transition(current_state, action)

        total_return += (gamma ** power) * reward
        power += 1

        if done:
            break

        current_state = next_state
        action = policy[current_state]

    return total_return

print("Policy:", policy)
print("State Value of S0:", state_value_function("S0"))
print("Action Value of S0 for Up:", action_value_function("S0", "up"))
```
