```python
M=input("enter number of machines")
M=int(M)
machines=[]
for i in range(M):
    name=input("enter machine name")
    machines.append(name)
print(machines)
```

```python
import numpy as np

N_play=5
N=np.zeros((M,N_play))
R=np.zeros((M,N_play))
Q=np.zeros((M,N_play))
UCB=[0,0,0]
Q_old=0

for machine in machines:
    print(f"Play with machine {machine}")
    r=input("enter reward")
    r=int(r)
    id=machines.index(machine)

    N[id,0]=1
    R[id,0]=r
    Q[id,0]=Q[id,0]+(R[id,0]-Q[id,0])/N[id,0]

print(N)
print(R)
print(Q)
```

```python
import math as mt
import numpy as np

play=1
c=2
t=3

UCB = np.zeros(M)

while(play<N_play):
    t=t+play

    for id in range(M):
        UCB[id]=Q[id,play-1]+c*mt.sqrt(mt.log(t)/np.sum(N[id,:]))

    id_max=np.argmax(UCB)

    print(f"play with machine {machines[id_max]}")
    # Fix: Validate input to ensure it's an integer
    while True:
        r_str = input("enter reward")
        if r_str.strip().isdigit() or (r_str.strip().startswith('-') and r_str.strip()[1:].isdigit()):
            r = int(r_str)
            break
        else:
            print("Invalid input. Please enter an integer for the reward.")

    N[id_max,play]=1
    R[id_max,play]=r

    Q[id_max,play]=Q[id_max,play-1]+(R[id_max,play]-Q[id_max,play-1])/np.sum(N[id_max,:])

    play=play+1

print(N)
print(R)
print(Q)
```
