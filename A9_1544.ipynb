Assignment1: Calculation of State and Action Values Using Bellman Expectation Equations for
stochastic policy and deterministic environment

```python

# Assignment 1: Algebraic solution of Bellman equations

gamma = 0.9

# Bellman equations:
# V0 = -1 + 0.81 * V1
# V1 = 89.9 + 0.09 * V0

# Substitute V1 into the V0 equation
V0 = 71.819 / 0.9271
V1 = 89.9 + 0.09 * V0

# Calculate action values
Q_S0_Down = -1 + gamma * V1
Q_S0_Right = -1
Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V0

# Verify state values using the stochastic policy
V0_check = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
V1_check = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

print("Assignment 1: Algebraic Solution")
print(f"V_pi(S0)       = {V0:.2f}")
print(f"V_pi(S1)       = {V1:.2f}")
print(f"Q_pi(S0, Down) = {Q_S0_Down:.2f}")
print(f"Q_pi(S0, Right)= {Q_S0_Right:.2f}")
print(f"Q_pi(S1, Right)= {Q_S1_Right:.2f}")
print(f"Q_pi(S1, Up)   = {Q_S1_Up:.2f}")

print("\nVerification:")
print(f"V0 check = {V0_check:.2f}")
print(f"V1 check = {V1_check:.2f}")
```

Assignment 2: Calculation of State and Action Values Using Bellman Expectation Equations for
stochastic environment and deterministic policy

```python

# Assignment 2: Iterative Policy Evaluation
# Deterministic policy, stochastic environment

gamma = 0.9
theta = 0.000001

# Initialize state values
V0 = 0.0
V1 = 0.0
iterations = 0

# Iterative policy evaluation
while True:
    # Use old values to update both states
    new_V0 = 0.1 * (-1 + gamma * 0) + \
             0.9 * (-1 + gamma * V1)

    new_V1 = 0.9 * (100 + gamma * 0) + \
             0.1 * (-1 + gamma * V0)

    delta = max(abs(new_V0 - V0),
                abs(new_V1 - V1))

    V0 = new_V0
    V1 = new_V1
    iterations += 1

    if delta < theta:
        break

# Calculate action values after convergence
Q_S0_Down = 0.1 * (-1 + gamma * 0) + \
            0.9 * (-1 + gamma * V1)

Q_S0_Right = -1 + gamma * 0

Q_S1_Right = 0.9 * (100 + gamma * 0) + \
             0.1 * (-1 + gamma * V0)

Q_S1_Up = -1 + gamma * V0

# Display results
print("Assignment 2: Iterative Policy Evaluation")
print("Number of iterations:", iterations)
print(f"V_pi(S0)        = {V0:.2f}")
print(f"V_pi(S1)        = {V1:.2f}")
print(f"Q_pi(S0, Down)  = {Q_S0_Down:.2f}")
print(f"Q_pi(S0, Right) = {Q_S0_Right:.2f}")
print(f"Q_pi(S1, Right) = {Q_S1_Right:.2f}")
print(f"Q_pi(S1, Up)    = {Q_S1_Up:.2f}")

# Verify policy consistency
print("\nPolicy Consistency Checks:")
print("V(S0) = Q(S0, Down):",
      abs(V0 - Q_S0_Down) < theta)
print("V(S1) = Q(S1, Right):",
      abs(V1 - Q_S1_Right) < theta)
```
