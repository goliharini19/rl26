# MODIFIED E-GREEDY

```python
M=input("enter number of machines")
M=int(M)

machines=[]

for i in range(M):
    name=input("enter machine name")
    machines.append(name)

print(machines)
```

```python
import numpy as np

N_plays = 3

N = np.zeros((M, N_plays))
R = np.zeros((M, N_plays))
Q = np.zeros((M, N_plays))

for machine in machines:
    print(f"Play with machine {machine}")

    reward = input("enter reward")
    reward = int(reward)

    id = machines.index(machine)

    N[id, 0] = 1
    R[id, 0] = reward

    Q[id, 0] = Q[id, 0] + (R[id, 0] - Q[id, 0]) / N[id, 0]

print(R)
print(N)
print(Q)
```

```python
import random

eps = 0.2
play = 1

while(play <= N_plays - 1):

    r = random.uniform(0, 1)

    id_max = np.argmax(Q[:, play - 1])

    if(r < eps):
        print("Exploration")

        ids = [i for i in range(M) if i != id_max] # except max value ramdomly choose any (modified e-greedy),whereas in E-greedy max value is also included

        id = random.choice(ids)

        print(f"Now select machine {machines[id]}")

        reward = input("enter reward")
        reward = int(reward)

        Q[:, play] = Q[:, play - 1]

        R[id, play] = reward
        N[id, play] = 1

        Q[id, play] = Q[id, play - 1] + (
            R[id, play] - Q[id, play - 1]
        ) / np.sum(N[id, :])

    else:
        print("Exploitation")

        id = id_max

        print(f"Now select machine {machines[id]}")

        reward = input("enter reward")
        reward = int(reward)

        Q[:, play] = Q[:, play - 1]

        R[id, play] = reward
        N[id, play] = 1

        Q[id, play] = Q[id, play - 1] + (
            R[id, play] - Q[id, play - 1]
        ) / np.sum(N[id, :])

    play = play + 1
    print(R)
    print(N)
    print(Q)
```

```python
# ε-Greedy rewards
total_reward = np.sum(R[:, 1:])
number_of_plays = np.sum(N[:, 1:])

average_reward = total_reward / number_of_plays

print("=" * 50)
print("Total ε-Greedy Reward =", int(total_reward))
print("Average ε-Greedy Reward =", round(average_reward, 2))
```

# Calculate total reward the player received

```python
total_reward = np.sum(R)
total_count = np.sum(N)

average_reward = total_reward / total_count

print("Total Reward =", int(total_reward))
print("Average Reward =", round(average_reward, 2))
```
