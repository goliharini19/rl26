Assignment 1: Calculate Undiscounted and Discounted Returns

```python
n = int(input("Enter the number of time steps: "))
rewards = []
for i in range(n):
    reward = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(reward)
undiscounted_return = sum(rewards)
gamma = float(input("Enter the discount factor (gamma): "))
discounted_return = 0
for i in range(n):
    discounted_return += rewards[i] * (gamma ** i)
print("\n----- Entered Reward Values -----")
for i in range(n):
    print(f"Time Step {i + 1}: {rewards[i]}")
print("\n----- Undiscounted Return -----")
print("Undiscounted Return =", undiscounted_return)
print("\n----- Discounted Return -----")
print("Discounted Return =", discounted_return)
```

```python
%%writefile grid_env_det.py

state_space = ["S0", "S1", "S3", "S4", "G"]

action_space = ["up", "down", "left", "right"]

discount_factor = 0.9

reward_function = {
    "normal": -1,
    "goal": 10,
    "obstacle": -5
}

transition_function = {
    ("S0", "up"): "S0",
    ("S0", "down"): "S0",
    ("S0", "left"): "S0",
    ("S0", "right"): "S1",

    ("S1", "up"): "S1",
    ("S1", "down"): "S3",
    ("S1", "left"): "S0",
    ("S1", "right"): "G",

    ("S3", "up"): "S1",
    ("S3", "down"): "S3",
    ("S3", "left"): "X",
    ("S3", "right"): "S4",

    ("S4", "up"): "G",
    ("S4", "down"): "S4",
    ("S4", "left"): "S3",
    ("S4", "right"): "S4"
}

def transition(state, action):
    if state == "G":
        return "G", reward_function["goal"], True

    if (state, action) not in transition_function:
        raise ValueError("Invalid state or action")

    next_state = transition_function[(state, action)]

    if next_state == "X":
        return "X", reward_function["obstacle"], True

    if next_state == "G":
        return "G", reward_function["goal"], True

    return next_state, reward_function["normal"], False
```

```python
%%writefile grid_world_mdp.py
import grid_env_det
print("State Space:", grid_env_det.state_space)
print("Action Space:", grid_env_det.action_space)
print("Discount Factor:", grid_env_det.discount_factor)

state = input("Enter current state: ")
action = input("Enter action: ")

next_state, reward, done = grid_env_det.transition(state, action)

print("Next State:", next_state)
print("Reward:", reward)
print("Episode Terminated:", done)
```
