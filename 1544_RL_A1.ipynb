LAB ASSIGNMENT - 1: GREEDY ALGORITHM

# Task - 1 :

```python
M=int(input("Number of machines in casino"))
machines=[]
for m in range(M):
 name=input("enter name of the machine")
 machines.append(name)
print(machines)
Q=[]
for machine in machines:
 N=input(f"enter number of plays happend for {machine}")
 N=int(N)
 average_reward=0
 for play in range(N):
  reward=input(f"enter reward in {play} play")
  average_reward=average_reward+int(reward)
 average_reward=average_reward/N
 Q.append(average_reward)
 print(f"Average reward of {machine} is {average_reward}")
print("Average rewards",Q)
id=Q.index(max(Q))
print(f"Machine with maximum reward is {machines[id]}")
```
